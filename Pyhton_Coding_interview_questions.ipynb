```python
#Finding the commen letters in two string
def common_letters():
  str1 = input("first string: ")
  str2 = input("second string: ")
  s1= set(str1)
  s2= set(str2)
  lst = s1&s2
  print(lst)

common_letters()
```

```python
#count the frequency of words in a string
def count_words():
  str = input("input the string: ")
  li = str.split()
  d = {}
  for i in li:
    if i not in d.keys():
      d[i] = 0
    d[i] = d[i]+1
  print(d)
count_words()
```
